#### 2. Data Type - Strings (Basic)

```python
# Changing Case
name = "john Joe"
print(name.title())
print(name.upper())
print(name.lower())

# Stripping Whitespace
nm = ' Hello '
print('before strip:'+nm)
nm = nm.strip()
print('After strip:'+nm)
```

#### 3. Data Type - Numbers

```python
a, b = 10, 5
print(f'{a} + {b} = {a+b}')
print(f'{a} - {b} = {a-b}')
print(f'{a} * {b} = {a*b}')
print(f'{a} / {b} = {a/b}')
print(f'{a} ** {b} = {a**b}')
print(f'{a} // {b} = {a//b}')
print(f'{a} % {b} = {a%b}')

# Type Error
#print('My sister is'+b+'years older than me')

# Solution
print('My sister is'+str(b)+'years older than me')
print('My sister is',b,'years older than me')
print(f'My sister is {b} years older than me')
```
