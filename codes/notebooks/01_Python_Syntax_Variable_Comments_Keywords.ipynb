# Python Syntax, Variables, Comments, and Keywords

## 1. Python Syntax - "How Python Speaks"

### What syntax really means

Syntax is **the rulebook of the language** - how you write code so Python can parse, interpret, and execute it.

Python's value proposition:

- Minimal syntax
- Human-readable
- No ceremony, no boilerplate

### Key Syntax Principles

✅Indention LAW

Python uses **Indentation instead of** {}.

````python
x = 5
if x > 0:
    print("Positive")
else:
    print("Non-positive")
````

If indentation is wrong → **runtime failure**, not a warning.

This enforces:

- readability

- consistency

- fewer hidden bugs

Tell it like it is:

> Python doesn’t trust developers to format code properly — it enforces it.
>

---

✅ No semicolons (usually)

````python
x = 10
y = 20
````

Semicolons are optional but discouraged:

````python
x = 10; y = 20 # legal, unprofessional
````

✅ Statements end at line breaks

One instruction per line = clarity.

## 2. Variables — “Named References to Objects”

This is where beginners get confused.

### What a variable REALLY is

A variable is **not a box**.
It’s a **label pointing to an object in memory.**

````python
x = 10
````

- `10` → object
- `x` → reference to that object

Python is:

- **dynamically typed**
- **strongly typed**

### Dynamic typing

You don’t declare types explicitly.

````python
x = 10
x = "hello"
x = 3.14
````

Same variable, different objects.

Python figures it out **at runtime**.

---

### Naming rules (non-negotiable)

**✅ Valid:**

````python
count = 10
total_sum = 50
is_valid = True
````

**❌ Invalid:**

````python
2value = 10      # starts with digit
total-sum = 5   # hyphen not allowed
class = 10      # keyword
````

### Naming convention 


| **Use case** | **Convention** |
|:---|:---:|
| Variables | `snake_case` |
| Constants | `UPPER_CASE` |
| Classes | `PascalCase` |

````python
MAX_RETRIES = 3
user_name = "Python"
````

## 3. Comments — “Communication with Humans”

Python code is read **far more than it’s written**.

Comments exist for:

- clarity
- intent
- future maintainers


### Single-line comments

````python
# This is a comment
x = 10  # inline comment
````

Use comments to explain **why**, not **what**.

**❌ Bad:**


````python
x = x + 1  # increment x
````

**✅ Good:**

````python
x += 1  # retry counter increases after failure
````

---

### Multi-line comments

Python has **NO true multi-line comment syntax**.

This:


````python
"""
This looks like a comment
"""
````


is actually a **string literal**, not a comment.

It’s fine for:

* docstrings
* documentation

But **don’t abuse it**.

---

### Docstrings

Used for documentation:


````python
def add(a, b):
    """Returns the sum of two numbers."""
    return a + b
````

This is how:

* help()
* IDEs
* documentation tools work

## 4. Keywords — “Reserved Power Words”

Keywords are **reserved identifiers** that Python owns.

You:

* cannot rename them
* cannot override them
* cannot use them as variables

---

### Common Python keywords (must-know)

```text
False   None   True
and     as     assert
break   class  continue
def     elif   else
for     if     import
in      is     lambda
not     or     pass
return  try    except
while   with   yield
```

Example:

````python
if True:
    pass
````

**❌ Illegal:**

````python
if = 10
class = "Python"
````

---

### How to see keywords in Python


````python
import keyword
print(keyword.kwlist)
````

Professional habit:

> Never fight the language — learn its reserved words early.
